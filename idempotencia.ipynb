```python
# Contagens ANTES de rodar o pipeline pela segunda vez
tabelas = [
    "orders", "order_items", "payments", "returns", "return_items",
    "fiscal_invoices", "customers", "products", "product_variants",
    "categories", "brands", "locations", "employees"
]

print("=== CONTAGENS ANTES DA SEGUNDA EXECUÇÃO ===")
contagens_antes = {}
for t in tabelas:
    count = spark.sql(f"SELECT COUNT(*) as total FROM dev.bronze.sql_{t}").collect()[0]["total"]
    contagens_antes[t] = count
    print(f"dev.bronze.sql_{t}: {count} linhas")
```

```python
print("\n=== CONTAGENS DEPOIS DA SEGUNDA EXECUÇÃO ===")
print(f"{'Tabela':<25} {'Antes':>10} {'Depois':>10} {'Resultado':>15}")
print("-" * 65)

todos_ok = True
for t in tabelas:
    depois = spark.sql(f"SELECT COUNT(*) as total FROM dev.bronze.sql_{t}").collect()[0]["total"]
    antes = contagens_antes[t]
    status = "✅ IDEMPOTENTE" if antes == depois else "❌ DIVERGENTE"
    if antes != depois:
        todos_ok = False
    print(f"sql_{t:<20} {antes:>10} {depois:>10} {status:>15}")

print("\n" + ("✅ PIPELINE IDEMPOTENTE — todas as contagens são iguais" if todos_ok else "❌ FALHA DE IDEMPOTÊNCIA — verificar o modo de escrita"))
```
