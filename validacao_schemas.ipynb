```python
TABELAS_BRONZE = [
    "sql_orders", "sql_order_items", "sql_payments", "sql_returns",
    "sql_return_items", "sql_fiscal_invoices", "sql_customers",
    "sql_products", "sql_product_variants", "sql_categories",
    "sql_brands", "sql_locations", "sql_employees",
]

for tabela in TABELAS_BRONZE:
    resultado = spark.sql(f"DESCRIBE DETAIL dev.bronze.{tabela}").collect()[0]
    print(f"{tabela}: format={resultado['format']}, numFiles={resultado['numFiles']}")
```

```python
for tabela in TABELAS_BRONZE:
    print(f"\n=== {tabela} ===")
    spark.sql(f"DESCRIBE dev.bronze.{tabela}").show(truncate=False)
```
