```python
# # Recuperar credenciais do Databricks Secrets
jdbc_host = dbutils.secrets.get(scope="lh-nautical", key="jdbc-host")
jdbc_port = dbutils.secrets.get(scope="lh-nautical", key="jdbc-port")
jdbc_database = dbutils.secrets.get(scope="lh-nautical", key="jdbc-database")
jdbc_user = dbutils.secrets.get(scope="lh-nautical", key="jdbc-user")
jdbc_password = dbutils.secrets.get(scope="lh-nautical", key="jdbc-password")

# # Montar a URL de conexão JDBC para SQL Server
jdbc_url = f"jdbc:sqlserver://{jdbc_host}:{jdbc_port};databaseName={jdbc_database};encrypt=true;trustServerCertificate=true"

connection_properties = {
    "user": jdbc_user,
    "password": jdbc_password,
    "driver": "com.microsoft.sqlserver.jdbc.SQLServerDriver"
}

print("Configuração de conexão montada com sucesso.")
```

```python
df_teste = spark.read.jdbc(
    url=jdbc_url,
    table="customers",
    properties=connection_properties
)

df_teste.show(5)
print(f"Total de linhas lidas: {df_teste.count()}")
```

```python
df_orders = spark.read.jdbc(
    url=jdbc_url,
    table="orders",
    properties=connection_properties
)

print(f"Total de linhas em orders: {df_orders.count()}")
```

```python
#Anotando tempo de leitura

import time

inicio = time.time()
df_orders.count()
fim = time.time()

print(f"Tempo de leitura de orders: {fim - inicio:.2f} segundos")
```
